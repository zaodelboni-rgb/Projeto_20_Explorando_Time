---
### 📚 Tabela de Métodos Principais da Biblioteca `time`

Antes de começarmos, aqui está uma referência rápida dos comandos que mais vamos usar:

| Método | O que faz? | Exemplo |
|--------|------------|---------|
| `time.sleep(segundos)` | **Pausa** a execução do código pelo tempo especificado. | `time.sleep(2)` → pausa de 2 segundos. |
| `time.time()` | Retorna o **timestamp atual** (segundos desde 01/01/1970). Usado para medir intervalos. | `inicio = time.time()` |
| `time.ctime()` | Retorna a data e hora atual em formato de texto legível. | `time.ctime()` → `'Wed Oct 14 10:30:00 2026'` |
| `time.localtime()` | Retorna a hora local como uma **estrutura** (ano, mês, dia, hora, etc.). | `time.localtime().tm_hour` |
| `time.strftime(formato)` | **Formata** a data/hora para o padrão que você quiser. | `time.strftime('%H:%M:%S')` |
| `time.perf_counter()` | Versão mais precisa de `time.time()`. Ideal para medir performance. | `time.perf_counter()` |

> 💡 **Importante:** Antes de usar qualquer um desses comandos, você precisa importar a biblioteca com `import time`.

```python
# Célula 0: Importando a biblioteca (Rode isso primeiro!)
import time
print("Biblioteca 'time' importada com sucesso!")
```
